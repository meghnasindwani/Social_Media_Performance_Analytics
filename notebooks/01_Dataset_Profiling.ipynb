5. Missing Value Analysis


```python
for name, df in datasets.items():
    print(f"\n{'='*50}")
    print(name)
    print(f"{'='*50}")

    missing = pd.DataFrame({
        "Missing Values": df.isnull().sum(),
        "Missing Percentage": (
            df.isnull().sum() / len(df) * 100
        ).round(2)
    })

    missing = missing[missing["Missing Values"] > 0]

    display(missing.sort_values(
        by="Missing Percentage",
        ascending=False
    ))
```

### 9. Initial Observations and Findings

9.1 Dataset Overview
The project uses three publicly available Instagram datasets to study different aspects of social media performance and content characteristics.

- DS01 – Reels Analysis: Contains 300 Instagram Reel records with 40 attributes. It includes information about creators, followers, publication dates, content topics, delivery styles, engagement metrics, and researcher-assigned content quality scores. The publication dates range from 4 February 2026 to 1 October 2026.

- DS02 – Academic Libraries: Contains 1,681 Instagram post records from 120 academic library profiles, with 7 attributes. It includes publication dates, captions/messages, profile information, post interaction rates, post types, and content categories. The publication dates range from 8 October 2023 to 8 November 2023.

- DS03 – Instagram Metrics: Contains 119 records with 13 attributes. It includes impressions, traffic sources, likes, comments, shares, saves, profile visits, follows, captions, and hashtags.

Each dataset provides a different perspective on Instagram content and audience performance.
9.2 Data Quality Observations
Initial profiling revealed the following data quality issues:

- DS01: Several columns contain missing values. Pinned comment has 83% missing values, CTA has 73% missing values, and Comment bait has approximately 60.67% missing values. The Views column contains values represented in different formats, such as K and M, and is currently stored as an object data type. It will require numerical conversion during preprocessing.

- DS02: The dataset has very few missing values. Seven messages and three publication dates are missing. The post interaction rate also requires verification of its calculation method and units.

- DS03: No missing values were detected. However, 17 exact duplicate records were identified among the 119 records, leaving 102 distinct records.

No exact duplicate records were found in DS01 and DS02.
9.3 Important Categorical Patterns
The categorical analysis revealed several interesting patterns:

DS01 – Reels Analysis
- Science is the most frequently occurring topic, with 71 records.
- Creator is the most common source credibility category, with 197 records.
- Curiosity is the most frequent emotional appeal, appearing in 134 records.
- Talking Head is the most common delivery style, with 120 records.
- Slow editing pace appears in 219 records.
- Visual examples are present in 295 records.
Some categorical columns, particularly Type and Emotional Appeal, contain multiple combined characteristics and may require further investigation.

DS02 – Academic Libraries
- Single Image is the most common post type, with 1,047 records.
- Carousel appears in 377 records.
- Reel appears in 257 records.
- Information about library events is the most common content category, with 383 records.
- Library services is the second most common category, with 308 records.
DS03 – Instagram Metrics

The dataset primarily contains numerical performance metrics. Caption and Hashtags are the two text-based attributes and may provide opportunities for text-based analysis.
9.4 Potential Analytical Opportunities
Based on the initial profiling, the following analytical opportunities have been identified:

DS01 – Content Intelligence
- Investigate the relationship between content characteristics and views.
- Examine whether delivery styles, editing pace, subtitles, and visual elements are associated with content performance.
- Explore relationships between researcher-assigned content quality scores and engagement metrics.
- Compare performance across different content topics and creators.
DS02 – Content Category Intelligence
- Compare interaction rates across different post types.
- Investigate the relationship between content categories and interaction rates.
- Identify content categories that demonstrate relatively higher interaction rates.
- Explore differences in posting patterns across academic library profiles.
DS03 – Audience Action Intelligence
- Analyse the contribution of different traffic sources to impressions.
- Examine relationships between impressions and engagement actions such as likes, comments, shares, and saves.
- Investigate the relationship between profile visits and follows.
- Explore caption and hashtag characteristics in relation to audience actions.
These are potential research directions and will be evaluated further during exploratory data analysis and statistical analysis.
9.5 Limitations Identified
The initial profiling identified the following limitations:
- The three datasets cover different populations, periods, and performance metrics. They cannot be directly combined into one homogeneous performance dataset.
- DS01 contains substantial missingness in certain content-related attributes. Its attention retention potential, education value, and content execution quality scores are researcher-assigned measures, not actual Instagram watch-time or retention metrics.
- DS01 does not currently have a numerical Views column because values are stored in abbreviated text formats.
- DS02 focuses specifically on academic library profiles, limiting the generalisation of findings to other industries.
- DS02 does not provide individual likes, comments, shares, reach, or impressions. Its interaction-rate calculation needs verification.
- DS03 has no publication date, post ID, or account ID, limiting time-based and account-level analysis.
- DS03 contains 17 exact duplicate records that require investigation before preprocessing.
- The datasets may not represent all Instagram users, industries, or content formats.
Therefore, the analysis and recommendations developed from these datasets will be limited to the available observations and should not be treated as universally applicable to all Instagram accounts.
